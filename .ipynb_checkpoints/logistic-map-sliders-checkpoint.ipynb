```python
from matplotlib.pyplot import figure, rc
import numpy as np
import matplotlib.pyplot as plt
import time
from ipywidgets import interact
import ipywidgets as widgets
from IPython.display import display
from numba import njit, prange


def logistic_growth_function(r, x):
    return r*x*(1-x)
    # 
    # N' = r*N*(K-N)
    # Entdimensionalisierung: neue Variable x=N/K 
    # 
def sine_growth_function (r,x):
    return r*np.sin(np.pi*x)/4
    
def iterate_logistic_growth_function(n_iterations, r, x_0):
    if type(x_0) == float or type(x_0) == np.longdouble:
        values = np.zeros(n_iterations, dtype=np.longdouble)
    else:
        values = np.zeros((n_iterations,x_0.shape[-1]), dtype=np.longdouble)
    values[0] = x_0
    for i in range(1, n_iterations):
        values[i] = logistic_growth_function(r, values[i-1])
    return values

def iterate_sine_growth_function(n_iterations, r, x_0):
    if type(x_0) == float or type(x_0) == np.longdouble:
        values = np.zeros(n_iterations, dtype=np.longdouble)
    else:
        values = np.zeros((n_iterations,x_0.shape[-1]), dtype=np.longdouble)
    values[0] = x_0
    for i in range(1, n_iterations):
        values[i] = sine_growth_function(r, values[i-1])
    return values


def plot_cobweb(r=0.1, x0=0.2, tmax=100, growth_function=logistic_growth_function, n_times_function_applied=1):
    x = np.linspace(0,1,500)
    fig = plt.figure(figsize=(15,8), dpi=100)
    ax=fig.add_subplot(111)
    def final_map(r,x):
        result = growth_function(r,x)
        for i in range(n_times_function_applied-1):
            result = growth_function(r, result)
        return result
    ax.plot(x, final_map(r,x), c='#444444', lw=2)
    ax.plot(x,x, c='#444444', lw=2)

    px, py = np.empty((2, tmax+1, 2))
    px[0] = x0
    py[0] = 0
    for n in range(1, tmax, 2):
        px[n] = px[n-1]
        py[n] = final_map(r, px[n-1])
        px[n+1] = py[n]
        py[n+1] = py[n]

    ax.plot(px, py, c="b", alpha=0.7)
    ax.set_xlim((0,1))
    ax.set_ylim((0,1))
    ax.set_xlabel("Anfangspopulationsgröße x")
    ax.set_ylabel("Endwert Populationsgröße $rx(1-x)$")
    ax.set_title('$x_0 = {:.1}, r = {:.7}$'.format(x0*1.,r*1.))

def plot_orbit_diagram(r_values=(3.0, 3.86),r_value_distance=0.0001, x0=0.5, n_iterations_to_plot_from=100, n_iterations_to_plot_until=200, labelsize=0.001, figsize=(20,10), dpi=100, discrete_map=iterate_logistic_growth_function, color="tab:blue", ax=None):
    all_r_values = np.arange(*r_values,r_value_distance)
    
    scatter_points = np.zeros((all_r_values.shape[-1], -n_iterations_to_plot_from+n_iterations_to_plot_until))
    for i,r in enumerate(all_r_values):
        scatter_points[i] = discrete_map(n_iterations_to_plot_until, r, x0)[n_iterations_to_plot_from:n_iterations_to_plot_until]
    if ax == None:
        fig, ax = plt.subplots(figsize=figsize, dpi=dpi)
    
    ax.scatter(np.repeat(all_r_values, -n_iterations_to_plot_from+n_iterations_to_plot_until), scatter_points.flatten(), s=labelsize, marker=".", c=color)
    ax.set_xlim(r_values)
    ax.set_ylim(0,1)
    ax.set_xlabel("r")
    ax.set_ylabel("Werte von x für $t \\in ({:},{:})$".format(n_iterations_to_plot_from, n_iterations_to_plot_until))
    ax.set_title('$x_0 = {:.1}, r \\in ({:.4},{:.4})$'.format(x0*1.0,r_values[0]*1.,r_values[1]*1.))
    return ax

def plot_trajectory_old_function(r=3.0, x0=0.5, n_iterations=100, figsize=(20,10), dpi=100):
    plt.figure(figsize=figsize, dpi=dpi)
    x = np.arange(1, n_iterations+1)
    y = iterate_logistic_growth_function(n_iterations, r, np.longdouble(x0))
    
    plt.plot(x[:n_iterations], y[:n_iterations])
    plt.xlabel("Zeitschritte t")
    plt.ylabel("Populationsgröße x")
    plt.ylim(0,1)
    plt.xlim(0,n_iterations)
    plt.title('$x_0 = {:.1}, r = {:.4}$'.format(x0*1.,r*1.))
```

```python
from dataclasses import dataclass

@dataclass
class OrbitFrame:
    r_min: float
    r_max: float
    x_min: float
    x_max: float

    image: np.ndarray
import numpy as np

@njit(parallel=True)
def compute_density_image_numba(
        r_min,
        r_max,
        x_min,
        x_max,
        r_pixels=1200,
        x_pixels=800,
        n_iterations_to_plot_from=400,
        n_iterations_to_plot_until=600,
        x0=0.5):

    H = np.zeros((x_pixels, r_pixels), dtype=np.uint32)

    for j in prange(r_pixels):

        r = r_min + (r_max-r_min)*j/(r_pixels-1)

        x = x0

        for k in range(n_iterations_to_plot_until):

            x = r*x*(1.0-x)

            if k >= n_iterations_to_plot_from:

                if x_min <= x <= x_max:

                    y_idx = int(
                        (x-x_min)
                        /(x_max-x_min)
                        *(x_pixels-1)
                    )

                    H[y_idx, j] += 1

    return H

def build_zoom_frames(
        r_center,
        x_center,
        initial_r_width,
        initial_x_width,
        n_frames=8,
        zoom_factor=2):

    frames = []
    r_min_init = 3.4
    r_max_init = 4
    x_min_init = 0
    x_max_init = 1
    img = compute_density_image_numba(
        r_min_init,
        r_max_init,
        x_min_init,
        x_max_init
    )
    frames.append(
        OrbitFrame(
            r_min=r_min_init,
            r_max=r_max_init,
            x_min=x_min_init,
            x_max=x_max_init,
            image=img
        )
    )
    for k in range(n_frames):

        scale = zoom_factor**k

        r_width = initial_r_width / scale
        x_width = initial_x_width / scale

        r_min = r_center - r_width/2
        r_max = r_center + r_width/2

        x_min = x_center - x_width/2
        x_max = x_center + x_width/2

        img = compute_density_image_numba(
            r_min,
            r_max,
            x_min,
            x_max
        )

        frames.append(
            OrbitFrame(
                r_min=r_min,
                r_max=r_max,
                x_min=x_min,
                x_max=x_max,
                image=img
            )
        )

    return frames

frames = build_zoom_frames(
    r_center=3.854076,
    x_center=0.166343,
    initial_r_width=4.0,
    initial_x_width=1.0,
    n_frames=8
)
#%matplotlib widget

fig, ax = plt.subplots(figsize=(8,6))

current_frame = 0

im = ax.imshow(
    np.log1p(frames[0].image),
    origin="lower",
    aspect="auto",
    extent=[
        frames[0].r_min,
        frames[0].r_max,
        frames[0].x_min,
        frames[0].x_max
    ],
    cmap="inferno"
)

ax.set_xlabel("r")
ax.set_ylabel("x")

zoom_slider = widgets.FloatSlider(
    value=0,
    min=0,
    max=len(frames)-1,
    step=0.01,
    description="Zoom",
    continuous_update=True
)
current_frame_idx = -1

def update_zoom(z):

    global current_frame_idx

    frame_idx = int(np.floor(z))

    frame_idx = min(
        frame_idx,
        len(frames)-1
    )

    frame = frames[frame_idx]

    if frame_idx != current_frame_idx:

        im.set_data(
            np.log1p(frame.image)
        )

        im.set_extent([
            frame.r_min,
            frame.r_max,
            frame.x_min,
            frame.x_max
        ])

        current_frame_idx = frame_idx

    local_zoom = z - frame_idx

    next_idx = min(
        frame_idx + 1,
        len(frames)-1
    )

    next_frame = frames[next_idx]

    alpha = local_zoom

    r_width = (
        (1-alpha)*(frame.r_max-frame.r_min)
        +
        alpha*(next_frame.r_max-next_frame.r_min)
    )

    x_width = (
        (1-alpha)*(frame.x_max-frame.x_min)
        +
        alpha*(next_frame.x_max-next_frame.x_min)
    )

    ax.set_xlim(
        frame.r_min + (frame.r_max-frame.r_min-r_width)/2,
        frame.r_max - (frame.r_max-frame.r_min-r_width)/2
    )

    ax.set_ylim(
        frame.x_min + (frame.x_max-frame.x_min-x_width)/2,
        frame.x_max - (frame.x_max-frame.x_min-x_width)/2
    )

    fig.canvas.draw_idle()
widgets.interactive_output(
    update_zoom,
    {"z": zoom_slider}
)

display(zoom_slider)
```
